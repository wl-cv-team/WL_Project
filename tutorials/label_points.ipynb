# 점 라벨 찍기

영상 하나에서 인상 구간의 경계 프레임을 직접 찍는다. 이미지는 화면에만 표시하고 파일로 저장하지 않는다.

- event: `ready`, `transition`, `drive`, `catch`, `finish`
- `pass_no`: 1 또는 2. 같은 영상에 pass 2 기록이 생기면 그 영상의 pass 1은 수정할 수 없다.
- `completion`: `f`는 5개 event 모두, `h`는 `ready`, `transition`, `drive`만 찍는다.
- 같은 `(영상, event, pass)`를 다시 찍으면 새 값으로 바뀐다.


## 1. 설정

영상 id, 영상 경로, pass 번호, completion을 적는다.

```python
from pathlib import Path

from wl_snatch.labeling import LabelingSession, PointLabelStore

video_id = "영상_id를_여기에"
video_path = Path("../data/raw/영상_파일.mp4")
pass_no = 1
completion = "f"
store = PointLabelStore(Path("../data/annotations/point_labels.csv"))
```

## 2. 라벨링

슬라이더나 -1, +1 버튼으로 프레임을 옮기고, event를 고른 뒤 record를 누른다.

```python
session = LabelingSession(video_path, video_id, store, pass_no, completion)
session.show()
```
