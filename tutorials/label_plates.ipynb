# 원판 박스 라벨링

프레임마다 카메라에 가까운 쪽 원판 하나의 보이는 외곽에 박스를 그린다. 이미지는 화면에만 표시하고 파일로 저장하지 않는다.

- 일부가 가려졌거나 화면 밖으로 잘렸으면 보이는 부분만 감싼다. 가려진 부분을 추정해 넓히지 않는다.
- 이미지 위에서 마우스를 눌러 끌었다 놓으면 박스 초안이 생긴다. 드래그가 되지 않으면 left, top, right, bottom 슬라이더(0~1)로 조정한다.
- `save box`는 초안을 저장하고, `save empty`는 이 프레임에 가까운 쪽 원판이 없다는 확인을 저장한다. 저장하지 않은 프레임은 "아직 라벨링 안 함"이다.
- 같은 프레임에 다시 저장하면 새 값으로 바뀐다. 프레임을 옮기면 초안은 지워진다.
- 영상에 추출 범위 예외가 있으면(`data/annotations/plate_frame_ranges.csv`) 그 범위 밖 프레임에는 저장할 수 없다.


## 1. 설정

영상 id와 영상 경로를 적는다.

`display_width`로 화면에 표시하는 이미지 폭(화면 픽셀)을 바꿀 수 있다. 값이 크면 마우스로 더 세밀하게 찍을 수 있지만 화면 세로 공간이 더 필요하다(세로 영상은 높이가 폭의 약 16/9배).

```python
from pathlib import Path

from wl_snatch.plate_boxes import BoxLabelingSession, PlateBoxStore

video_id = "영상_id를_여기에"
video_path = Path("../data/raw/영상_파일.mp4")
display_width = 360
store = PlateBoxStore(
    Path("../data/annotations/plate_boxes"),
    Path("../data/annotations/plate_frame_ranges.csv"),
)
```

## 2. 라벨링

슬라이더나 -1, +1 버튼으로 프레임을 옮기고, 박스를 그린 뒤 `save box` 또는 `save empty`를 누른다.

```python
session = BoxLabelingSession(video_path, video_id, store, display_width=display_width)
session.show()
```
