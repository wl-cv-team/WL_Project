## 1. 설정

모델 이름(`yolo26`, `yolo11`, `yolov8`)과 영상 id를 적는다. 영상은 `<videos_dir>/<영상 id>.mp4`에서 읽는다. 영상은 저장소에 없으므로 각자 로컬 `data/raw/videos/`에 있어야 한다.

라벨된 영상 id(라벨된 프레임이 있는 영상): `a01_sn_f_01`, `a03_sn_f_01`, `a05_sn_f_01`, `a06_sn_h_01`, `b01_sn_f_01`, `b02_sn_h_02`, `b03_sn_f_01`, `b04_sn_f_01`, `b05_sn_f_01`, `b06_sn_f_01`, `c02_sn_f_01`, `c04_sn_f_01`, `c05_sn_f_01`, `c08_sn_f_01`.

처음 실행하면 모델 가중치를 현재 폴더(`tutorials/`)에 내려받는다. `*.pt`는 git에서 제외된다. 이 노트북을 저장해 `git status`에 변경이 생기면 커밋하지 말고 되돌린다.

```python
from pathlib import Path

import pandas as pd

model_name = "yolo26"
video_id = "영상_id를_여기에"
videos_dir = Path("../data/raw/videos")
boxes_dir = Path("../data/annotations/plate_boxes")
ranges_path = Path("../data/annotations/plate_frame_ranges.csv")
videos = pd.read_csv("../data/metadata/videos.csv")
split = pd.read_csv("../data/splits/person_split.csv")
```

## 2. 세션

모델을 불러와(처음에는 가중치를 내려받는다) 이 영상의 라벨된 프레임에서만 실행한다.

```python
from wl_snatch.coco_baseline import load_coco_model
from wl_snatch.config import DetectorConfig
from wl_snatch.detection_viewer import build_view_session

config = DetectorConfig(model_name=model_name)
model = load_coco_model(config)
session = build_view_session(
    config, model, video_id, videos_dir, boxes_dir, videos, split, ranges_path
)
```

## 3. 확인

슬라이더로 프레임을 옮기고, 클래스와 신뢰도 하한을 골라 본다.
